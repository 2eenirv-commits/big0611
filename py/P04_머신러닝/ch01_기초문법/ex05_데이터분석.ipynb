```python
# 리스트
lst = list(range(10))
print(lst)
print(lst * 2)

# result = []
# for 변수 in 반복가능객체:
#     result.append(변수 * 2)

result = []
for num in lst:
    result.append(num * 2)    
print(result)    
```

```python
import numpy as np
# 배열(NumPy)
arr = np.arange(10)
print(arr)
print(arr * 2)
```
