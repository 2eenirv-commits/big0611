```python
# 데이터 유형: 숫자(정수), [리스트], '문자열'
# 변수: numbers, i
# 연산자: 대입(=), 비교(<, ==)
# 제어문: 조건문(if~elif), 반복문(for~in)
# 내장 함수: range(), print(), str()
# f-string: '{변수} 문자열'

numbers = range(10)

for i in numbers:
    if i < 5:
        print('{}는 5보다 작습니다.'.format(i))
    elif i == 5:
        print('{}는 5와 같습니다.'.format(i))   
    else:
        print('{}는 5보다 큽니다.'.format(i))    
```
